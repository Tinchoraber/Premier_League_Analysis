# Análisis de la Premier League 1993-2023

## Introducción
Este proyecto analiza más de 30 temporadas de la Premier League con el objetivo 
de identificar qué factores determinan si un equipo termina en los primeros puestos 
o pelea el descenso.

### Herramientas utilizadas
- Python (Pandas, NumPy)
- Matplotlib y Seaborn
- Jupyter Notebooks

### Preguntas del análisis
1. ¿Quién gana más, el local o el visitante?
2. ¿Cómo evolucionaron los goles a lo largo de los años?
3. ¿Qué equipos dominaron históricamente?
4. ¿Qué factores separan a los equipos del Top 4 de los que descienden?

## Hallazgos Principales

### 1. Ventaja del local
El equipo local gana aproximadamente el **46% de los partidos**, 
mientras que el visitante solo gana el **26%**. 
Jugar de local es una ventaja significativa y constante en toda la historia de la Premier.

### 2. Evolución de los goles
El promedio de goles por partido fue mínimo entre **2006 y 2009** (2.45 goles).
Desde 2010 hay una tendencia clara que va creciendo, llegando al máximo histórico 
de **2.85 goles por partido en 2023**.

### 3. Impacto del COVID
En la temporada 2021, jugada **sin público**, la ventaja del local casi desapareció.
Las líneas de goles de local y visitante casi se tocaron, confirmando que 
**el público influye directamente en el rendimiento del equipo local**.

### 4. Dominancia histórica
**Manchester United** lidera históricamente en victorias y goles de local, 
seguido de cerca por Arsenal y Liverpool.

### 5. Top 4 vs Descenso
Los equipos del Top 4 en 2023 acumularon más de 70 puntos, producto de 
una gran cantidad de victorias y muy pocas derrotas. En cambio, los equipos 
que descendieron no llegaron a los 35 puntos, lo que refleja más derrotas 
que victorias a lo largo de toda la temporada.
La diferencia de goles también es un factor clave: los equipos del Top 4 
terminaron con diferencia positiva amplia, mientras que los que descendieron 
terminaron recibiendo muchos más goles de los que convirtieron.
En resumen, **ganar partidos de local, convertir más goles de los que se reciben 
y mantener la consistencia durante toda la temporada** son los tres factores 
que más determinan la posición final de un equipo en la Premier League.

## Conclusiones Finales

El análisis de más de 30 temporadas de la Premier League permite concluir que:

- **Jugar de local es una ventaja real y medible**: el equipo local gana 
casi el doble de partidos que el visitante, y esta ventaja se mantuvo 
constante durante toda la historia de la liga.

- **El público es un factor determinante**: en la temporada 2021 jugada 
sin público por el COVID, la ventaja del local casi desapareció, 
confirmando que el apoyo de la hinchada influye directamente 
en el rendimiento del equipo.

- **El fútbol moderno genera más goles**: la tendencia desde 2010 
es sostenida, llegando al máximo histórico de 2.85 goles por partido en 2023.

- **La consistencia define a los campeones**: Manchester United, Arsenal 
y Liverpool dominaron históricamente gracias a su rendimiento 
sostenido de local durante décadas.

- **Los tres factores que determinan si un equipo entra al Top 4 
o desciende son**: ganar partidos de local, tener una diferencia 
de goles positiva y mantener la consistencia durante toda la temporada.
